## 2. Data Loading and Preprocessing

```python
df = load_data('../data/dummy_data.csv')
df.head()
```

## 3. Feature Engineering

```python
df_features = add_features(df)
print(f"Original number of columns: {len(df.columns)}")
print(f"Number of columns after adding features: {len(df_features.columns)}")
df_features.head()
```

## 4. Training a Transformer Model

```python
# Define target variable (1 if price goes up, 0 otherwise)
df_features['target'] = (df_features['close'].shift(-1) > df_features['close']).astype(int)
df_features.dropna(inplace=True)

X = df_features.drop('target', axis=1)
y = df_features['target']

transformer_model = train_transformer(X, y, epochs=5) # Using fewer epochs for demonstration
```

## 5. Backtesting the Transformer Model

```python
predictions = predict_transformer(transformer_model, X)
signals = pd.Series(0, index=X.index)
signals[predictions == 1] = 1 # Buy
signals[predictions == 0] = 2 # Sell

backtester = Backtester(data=df.loc[X.index], signals=signals)
metrics = backtester.run()

print("Backtesting Metrics:")
for key, value in metrics.items():
    print(f"- {key}: {value}")

backtester.plot_performance()
```
